```python
# ============================================================
# TASK 2: Train & evaluate ML models
# Drug Review Sentiment and Side-Effect Mining (UCI Drug Review)
# Models: Complement Naive Bayes, Logistic Regression, Linear SVM
# For research/education only - not medical advice.
# ============================================================
import os, re, io, json, html, time, zipfile, warnings
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests, joblib
from IPython.display import Markdown, display
from sklearn.base import clone
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import ComplementNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.model_selection import (train_test_split, GridSearchCV, StratifiedKFold,
                                     cross_val_score, GroupShuffleSplit)
from sklearn.preprocessing import label_binarize
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score, precision_score,
                             recall_score, roc_auc_score, matthews_corrcoef, confusion_matrix,
                             classification_report)
warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ---------------- SETTINGS ----------------
OWN_FILE      = None     # optional path, e.g. "/content/drugsComTrain_raw.tsv" (tab or comma separated)
MAX_ROWS      = 50000    # reviews to use (full data is about 215K; more rows = slower)
TEST_SIZE     = 0.20
RANDOM_STATE  = 42
DO_TUNING     = True     # small GridSearchCV per model
TUNE_SAMPLE   = 20000    # rows used for the grid search (final models are refit on the full train set)
CV_FOLDS      = 5        # stratified k-fold cross-validation on the train set
N_BOOT        = 500      # bootstrap resamples for test-set confidence intervals
OUT_DIR       = "results"
UCI_URL = "https://archive.ics.uci.edu/static/public/462/drug+review+dataset+drugs+com.zip"
os.makedirs(OUT_DIR, exist_ok=True)

# ============ 1. LOAD DATA ============
import sys, subprocess

OWN_FILE = None   # a path, or a list of paths, e.g. ["/content/drugsComTrain_raw.tsv", "/content/drugsComTest_raw.tsv"]

def read_any(src, name=""):
    """Read a tsv/csv from a path or raw bytes."""
    sep = "\t" if str(name or src).lower().endswith(".tsv") else None
    buf = io.BytesIO(src) if isinstance(src, (bytes, bytearray)) else src
    return pd.read_csv(buf, sep=sep, engine="python" if sep is None else "c")

def load_drug_reviews():
    # 0) your own file(s)
    if OWN_FILE:
        paths = [OWN_FILE] if isinstance(OWN_FILE, str) else list(OWN_FILE)
        return pd.concat([read_any(p) for p in paths], ignore_index=True)

    # 1) direct download (several known locations)
    urls = ["https://archive.ics.uci.edu/ml/machine-learning-databases/00462/drugsCom_raw.zip", UCI_URL]
    for u in urls:
        try:
            r = requests.get(u, timeout=120); r.raise_for_status()
            z = zipfile.ZipFile(io.BytesIO(r.content))
            parts = [read_any(z.read(n), n) for n in z.namelist() if n.lower().endswith((".tsv", ".csv"))]
            if parts:
                print("Downloaded from:", u); return pd.concat(parts, ignore_index=True)
        except Exception as ex:
            print(f"  failed: {u[:75]}... ({type(ex).__name__})")

    # 2) official UCI Python package
    try:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "ucimlrepo"], check=True)
        from ucimlrepo import fetch_ucirepo
        d = fetch_ucirepo(id=462)
        data = pd.concat([d.data.features, d.data.targets], axis=1)
        if {"review", "rating"} <= set(data.columns):
            print("Downloaded via ucimlrepo"); return data
    except Exception as ex:
        print(f"  ucimlrepo failed ({type(ex).__name__}: {str(ex)[:60]})")

    # 3) ask you to upload the files (Colab)
    from google.colab import files
    print("\nAuto-download failed. Select drugsComTrain_raw.tsv (and drugsComTest_raw.tsv if you have it):")
    up = files.upload()
    if not up: raise RuntimeError("No file uploaded")
    return pd.concat([read_any(b, n) for n, b in up.items()], ignore_index=True)

df, USING_DEMO = None, False
try:
    df = load_drug_reviews()
except Exception as ex:
    print(f"Could not load the real dataset ({type(ex).__name__}: {str(ex)[:80]})")
    print(">>> USING A SMALL SYNTHETIC DEMO DATASET. Numbers below are NOT real results. <<<")
    df, USING_DEMO = make_demo_data(), True

# ============ 2. CLEAN TEXT (stop words KEPT so negations survive) ============
def clean_review(s):
    s = html.unescape(str(s))                       # &#039; -> '
    s = s.replace("\u2019", "'").replace("\r", " ").replace("\n", " ")
    s = re.sub(r"http\S+|www\.\S+", " ", s)
    s = s.strip().strip('"').strip("'")
    return re.sub(r"\s+", " ", s).strip()

n0 = len(df)
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
df = df.dropna(subset=["review", "rating"])
df = df[df["rating"].between(1, 10)]
df["review"] = df["review"].apply(clean_review)
df = df.drop_duplicates(subset="review")
df = df[df["review"].str.split().str.len() >= 5]
df["drugName"] = df["drugName"].fillna("Unknown").astype(str).str.strip()
df = df.reset_index(drop=True)
print(f"Cleaning: {n0:,} -> {len(df):,} rows")

# ============ 3. SENTIMENT LABELS (from rating) + CLASS BALANCE ============
LABELS = ["negative", "neutral", "positive"]
L2I = {l: i for i, l in enumerate(LABELS)}
df["sentiment"] = pd.cut(df["rating"], bins=[0, 4, 6, 10], labels=LABELS).astype(str)
df["y"] = df["sentiment"].map(L2I)
dist = df["sentiment"].value_counts().reindex(LABELS)
print((dist / len(df)).round(3).to_string())
fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
dist.plot(kind="bar", ax=ax[0], color=["tomato", "goldenrod", "seagreen"]); ax[0].set_title("Class distribution")
sns.histplot(df["review"].str.split().str.len().clip(upper=300), bins=40, ax=ax[1]); ax[1].set_title("Review length (words, clipped at 300)")
plt.tight_layout(); plt.show()

# ============ 4. TRAIN / TEST SPLIT (stratified) ============
X_train, X_test, y_train, y_test = train_test_split(
    df["review"], df["y"], test_size=TEST_SIZE, stratify=df["y"], random_state=RANDOM_STATE)
print(f"Train: {len(X_train):,} | Test: {len(X_test):,}")
print("Train class share:", y_train.value_counts(normalize=True).sort_index().round(3).tolist())
print("Test  class share:", y_test.value_counts(normalize=True).sort_index().round(3).tolist())

# ============ 5. MODELS + HYPERPARAMETER TUNING ============
# TF-IDF is INSIDE the pipeline, so it is fitted on training data only (no leakage).
def make_pipe(clf):
    return Pipeline([("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=3, max_features=50000, sublinear_tf=True)),
                     ("clf", clf)])

CANDIDATES = {
    "Complement Naive Bayes": (make_pipe(ComplementNB()),
                               {"clf__alpha": [0.1, 0.3, 1.0]}),
    "Logistic Regression":    (make_pipe(LogisticRegression(max_iter=1000, class_weight="balanced")),
                               {"clf__C": [0.5, 1, 4, 10]}),
    "Linear SVM":             (make_pipe(LinearSVC(class_weight="balanced")),
                               {"clf__C": [0.05, 0.1, 0.5, 1]}),
}

if len(X_train) > TUNE_SAMPLE:
    Xs, _, ys, _ = train_test_split(X_train, y_train, train_size=TUNE_SAMPLE, stratify=y_train, random_state=RANDOM_STATE)
else:
    Xs, ys = X_train, y_train

models, tuning_rows, fit_times = {}, [], {}
for name, (pipe, grid) in CANDIDATES.items():
    if DO_TUNING:
        gs = GridSearchCV(pipe, grid, scoring="f1_macro", cv=3, n_jobs=-1)
        gs.fit(Xs, ys)
        best_params = gs.best_params_
        tuning_rows.append({"model": name, "best_params": best_params, "tuning_macro_F1": round(gs.best_score_, 4)})
        pipe = clone(pipe).set_params(**best_params)
    t0 = time.time(); pipe.fit(X_train, y_train); fit_times[name] = time.time() - t0
    models[name] = pipe
    print(f"{name:24s} fitted in {fit_times[name]:.1f}s")
if tuning_rows:
    display(pd.DataFrame(tuning_rows))

# ============ 6. TEST-SET EVALUATION ============
# Main metric = macro-F1 (classes are imbalanced, so accuracy alone is misleading)
def get_scores(pipe, X):
    return pipe.predict_proba(X) if hasattr(pipe, "predict_proba") else pipe.decision_function(X)

def macro_auc(y_true, S):
    Yb = label_binarize(y_true, classes=list(range(len(LABELS))))
    return float(np.mean([roc_auc_score(Yb[:, k], S[:, k]) for k in range(len(LABELS))]))

preds, scores, rows = {}, {}, []
for name, pipe in models.items():
    t0 = time.time(); p = pipe.predict(X_test); pred_t = time.time() - t0
    S = get_scores(pipe, X_test); preds[name], scores[name] = p, S
    rows.append({"model": name,
                 "accuracy": accuracy_score(y_test, p),
                 "balanced_acc": balanced_accuracy_score(y_test, p),
                 "macro_precision": precision_score(y_test, p, average="macro"),
                 "macro_recall": recall_score(y_test, p, average="macro"),
                 "macro_F1": f1_score(y_test, p, average="macro"),
                 "weighted_F1": f1_score(y_test, p, average="weighted"),
                 "macro_ROC_AUC": macro_auc(y_test, S),
                 "MCC": matthews_corrcoef(y_test, p),
                 "fit_time_s": fit_times[name], "predict_time_s": pred_t})
results = pd.DataFrame(rows).set_index("model").sort_values(["macro_F1", "macro_ROC_AUC"], ascending=False)
print("\n=== TEST-SET METRICS ===")
display(results.round(4))

best_name = results.index[0]
print("Best model by macro-F1:", best_name, "\n")
for name in results.index:
    print("=" * 70, "\n", name, "\n" + "=" * 70)
    print(classification_report(y_test, preds[name], target_names=LABELS, digits=3))

fig, ax = plt.subplots(1, len(models), figsize=(5 * len(models), 4))
for a, name in zip(np.atleast_1d(ax), results.index):
    cm = confusion_matrix(y_test, preds[name], normalize="true")
    sns.heatmap(cm, annot=True, fmt=".2f", cmap="Blues", vmin=0, vmax=1, xticklabels=LABELS, yticklabels=LABELS, ax=a, cbar=False)
    a.set_title(f"{name}\nmacro-F1 {results.loc[name, 'macro_F1']:.3f}"); a.set_xlabel("Predicted"); a.set_ylabel("True")
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/confusion_matrices.png", dpi=150); plt.show()

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5))
results[["accuracy", "balanced_acc", "macro_F1", "weighted_F1", "macro_ROC_AUC", "MCC"]].T.plot(kind="bar", ax=ax[0])
ax[0].set_title("Overall metrics by model"); ax[0].set_ylim(0, 1); ax[0].tick_params(axis="x", rotation=30); ax[0].legend(fontsize=8)
pc = pd.DataFrame({name: f1_score(y_test, preds[name], average=None) for name in results.index}, index=LABELS)
pc.plot(kind="bar", ax=ax[1]); ax[1].set_title("Per-class F1"); ax[1].set_ylim(0, 1); ax[1].tick_params(axis="x", rotation=0)
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/metric_comparison.png", dpi=150); plt.show()

# ============ 7. IS THE DIFFERENCE REAL? CROSS-VALIDATION + BOOTSTRAP ============
skf = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)
cv_rows = []
for name, pipe in models.items():
    s = cross_val_score(clone(pipe), X_train, y_train, cv=skf, scoring="f1_macro", n_jobs=-1)
    cv_rows.append({"model": name, "cv_macro_F1_mean": s.mean(), "cv_macro_F1_std": s.std()})
cv_df = pd.DataFrame(cv_rows).set_index("model").sort_values("cv_macro_F1_mean", ascending=False)
print("\n=== CROSS-VALIDATION (macro-F1) ===")
display(cv_df.round(4))

rng = np.random.RandomState(RANDOM_STATE)
y_arr = np.asarray(y_test)
boot = {name: [] for name in models}
for _ in range(N_BOOT):
    idx = rng.randint(0, len(y_arr), len(y_arr))
    for name in models:
        boot[name].append(f1_score(y_arr[idx], preds[name][idx], average="macro"))
boot = {k: np.array(v) for k, v in boot.items()}

ci_rows = []
for name in results.index:
    lo, hi = np.percentile(boot[name], [2.5, 97.5])
    row = {"model": name, "macro_F1": results.loc[name, "macro_F1"], "CI95_low": lo, "CI95_high": hi}
    if name != best_name:
        d = boot[best_name] - boot[name]
        dlo, dhi = np.percentile(d, [2.5, 97.5])
        row.update({"diff_vs_best_mean": 0.0 - d.mean(), "diff_CI95_low": 0.0 - dhi, "diff_CI95_high": 0.0 - dlo,
                    "gap_is_significant": bool(dlo > 0)})
    ci_rows.append(row)
ci_df = pd.DataFrame(ci_rows).set_index("model")
print("\n=== BOOTSTRAP 95% CONFIDENCE INTERVALS ===")
display(ci_df.round(4))

# ============ 8. ROBUSTNESS: TEST ON UNSEEN DRUGS ============
gss = GroupShuffleSplit(n_splits=1, test_size=TEST_SIZE, random_state=RANDOM_STATE)
gi_tr, gi_te = next(gss.split(df["review"], df["y"], groups=df["drugName"]))
Xg_tr, Xg_te = df["review"].iloc[gi_tr], df["review"].iloc[gi_te]
yg_tr, yg_te = df["y"].iloc[gi_tr], df["y"].iloc[gi_te]
g_rows = []
for name, pipe in models.items():
    m = clone(pipe).fit(Xg_tr, yg_tr); p = m.predict(Xg_te)
    g_rows.append({"model": name, "grouped_macro_F1": f1_score(yg_te, p, average="macro"),
                   "grouped_accuracy": accuracy_score(yg_te, p)})
group_df = pd.DataFrame(g_rows).set_index("model").join(results[["macro_F1"]].rename(columns={"macro_F1": "random_split_macro_F1"}))
print("\n=== UNSEEN-DRUG ROBUSTNESS ===")
display(group_df.round(4))

# ============ 9. WHAT DID THE MODELS LEARN? ERROR ANALYSIS ============
lr = models["Logistic Regression"]
vocab = np.array(lr.named_steps["tfidf"].get_feature_names_out())
coef = lr.named_steps["clf"].coef_
print("\n=== TOP WORDS PER CLASS (Logistic Regression) ===")
for k, lab in enumerate(LABELS):
    print(f"Top words for {lab.upper():9s}:", ", ".join(vocab[np.argsort(-coef[k])[:12]]))

cm = confusion_matrix(y_test, preds[best_name])
off = [(cm[i, j], LABELS[i], LABELS[j]) for i in range(3) for j in range(3) if i != j]
print(f"\nMost common confusions for {best_name} (count, true -> predicted):")
for n, t, p in sorted(off, reverse=True)[:3]: print(f"  {n:5d}  {t} -> {p}")

err = pd.DataFrame({"review": X_test.values, "true": [LABELS[i] for i in y_test], "pred": [LABELS[i] for i in preds[best_name]]})
err = err[err.true != err.pred].sample(min(8, (err.true != err.pred).sum()), random_state=RANDOM_STATE)
pd.set_option("display.max_colwidth", 140)
print("\nSample of misclassified reviews:")
display(err)

# ============ 10. WHY DOES ONE MODEL PERFORM BETTER? (written from YOUR numbers) ============
WHY = {
 "Linear SVM": "its max-margin objective works very well on sparse, high-dimensional TF-IDF features, it learns weights jointly (so correlated unigrams/bigrams are not double-counted), and class weighting helps the minority neutral class.",
 "Logistic Regression": "it learns discriminative weights jointly on correlated unigram/bigram features (so, unlike Naive Bayes, evidence is not double-counted), regularisation controls the very high feature count, and class weighting helps the minority neutral class.",
 "Complement Naive Bayes": "with strong, class-specific vocabulary and limited tuning, its simple word-count model is competitive and robust; its independence assumption did not hurt as much as expected on this data.",
}
others = [m for m in results.index if m != best_name]
lines = [f"**Best model: {best_name}** (macro-F1 {results.loc[best_name,'macro_F1']:.3f}, accuracy {results.loc[best_name,'accuracy']:.3f}, "
         f"macro ROC-AUC {results.loc[best_name,'macro_ROC_AUC']:.3f}).", ""]
for o in others:
    r = ci_df.loc[o]
    verdict = ("the gap is statistically meaningful (95% bootstrap interval of the difference excludes 0)"
               if r["gap_is_significant"] else "the gap is NOT statistically significant (interval includes 0), so the two are effectively tied")
    lines.append(f"- vs {o}: macro-F1 {results.loc[o,'macro_F1']:.3f} (difference {r['diff_vs_best_mean']:+.3f}); {verdict}.")
per_cls = {m: f1_score(y_test, preds[m], average=None) for m in results.index}
weak_txt = ", ".join(f"{m}: weakest class = {LABELS[int(np.argmin(v))]} (F1 {v.min():.2f})" for m, v in per_cls.items())
cv_rank = ", ".join(f"{m} {cv_df.loc[m, 'cv_macro_F1_mean']:.3f}" for m in cv_df.index)
cv_line = (f"Cross-validation agrees on the ranking: {cv_rank}." if list(cv_df.index)[0] == best_name
           else f"Note: cross-validation ranks {cv_df.index[0]} first, so the ranking is not fully stable; treat the top models as close.")
lines += ["", cv_line,
          f"Per-class view: {weak_txt}.",
          f"On unseen drugs, macro-F1 for {best_name} is {group_df.loc[best_name,'grouped_macro_F1']:.3f} vs {group_df.loc[best_name,'random_split_macro_F1']:.3f} on the random split.",
          ""]
any_sig = any(bool(ci_df.loc[o, "gap_is_significant"]) for o in others)
if any_sig:
    lines.append(f"**Why {best_name} wins here:** {WHY[best_name]}")
else:
    lines.append(f"**No statistically clear winner.** {best_name} ranks first on the point estimates, but the bootstrap intervals overlap, "
                 "so the honest conclusion is that the models perform about equally on this data. In that case choose on secondary criteria: "
                 "training/prediction speed, probability outputs, and interpretability (see the trade-offs above). "
                 f"If {best_name} is kept, the usual reason is: {WHY[best_name]}")
conclusion = "\n".join(lines)
if USING_DEMO: conclusion = "**WARNING: these numbers come from SYNTHETIC demo data, not the real UCI dataset.**\n\n" + conclusion
display(Markdown("## Conclusion\n\n" + conclusion))

# ============ 11. SAVE RESULTS FOR SUBMISSION ============
results.round(4).to_csv(f"{OUT_DIR}/model_comparison.csv")
pc.round(4).to_csv(f"{OUT_DIR}/per_class_f1.csv")
cv_df.round(4).to_csv(f"{OUT_DIR}/cross_validation.csv")
ci_df.round(4).to_csv(f"{OUT_DIR}/bootstrap_ci.csv")
group_df.round(4).to_csv(f"{OUT_DIR}/unseen_drug_robustness.csv")
joblib.dump(models[best_name], f"{OUT_DIR}/best_model.joblib")
with open(f"{OUT_DIR}/results_summary.md", "w") as f:
    f.write("# Results: Drug Review Sentiment (UCI Drug Review)\n\n")
    f.write(f"Rows used: {len(df):,} | train {len(X_train):,} | test {len(X_test):,} | demo data: {USING_DEMO}\n\n")
    f.write("## Test-set metrics\n\n" + results.round(4).to_markdown() + "\n\n")
    f.write("## Cross-validation (macro-F1)\n\n" + cv_df.round(4).to_markdown() + "\n\n")
    f.write("## Bootstrap 95% CI\n\n" + ci_df.round(4).to_markdown() + "\n\n")
    f.write("## Unseen-drug robustness\n\n" + group_df.round(4).to_markdown() + "\n\n")
    f.write("## Conclusion\n\n" + conclusion + "\n")
print("Saved to ./%s:" % OUT_DIR, sorted(os.listdir(OUT_DIR)))

# Colab: bundle everything into one zip to download
try:
    import shutil; shutil.make_archive("task2_results", "zip", OUT_DIR)
    from google.colab import files; files.download("task2_results.zip")
except Exception:
    pass
```
